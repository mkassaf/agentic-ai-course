## Ask another agent to check the proposal

An architecture proposer can delegate a policy check to a reviewer. A2A provides agent discovery and message exchange; our reviewer applies the same fictional rule deterministically.

`review_over_a2a` is a **course helper**, not an SDK method. It wraps the official SDK's Agent Card discovery and A2A 1.0 JSON-RPC exchange inside one process. [Inspect the helper](https://github.com/mkassaf/agentic-ai-course/blob/main/coursekit/protocol_examples.py). No public server or model key is needed.

```python
from coursekit.protocol_examples import review_over_a2a
case = {"team_size": 3, "independent_deploys": False}
agent, report = await review_over_a2a(case, "modular monolith")
print(f"{agent}: {report}")
assert report == "PASS: course policy expects modular monolith."
```

**Expected output:** `Architecture reviewer: PASS: course policy expects modular monolith.`
