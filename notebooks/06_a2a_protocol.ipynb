## Ask another agent to review the proposal

**So far:** one program did all the work, calling a model and its own tools.

**New:** *delegation*. Our code sends its proposal to a separate reviewer agent over the **Agent2Agent (A2A)** protocol and gets back a PASS or FAIL report.

The course helper `review_over_a2a` performs three real A2A steps with the official SDK, all inside this notebook:

1. **Start** a reviewer agent that publishes an *Agent Card* with its name, description, and skills.
2. **Discover** the reviewer by fetching that card, just as a client does for an agent it has not used before.
3. **Send** a message that contains the case and the proposal. The reviewer applies the course policy and replies.

The reviewer is plain Python. It uses no LLM and needs no model key. `review_over_a2a` is a course helper, not an SDK method: [read its code](https://github.com/mkassaf/agentic-ai-course/blob/main/coursekit/protocol_examples.py).

## Send a proposal to the reviewer

`agent` is the name read from the discovered Agent Card, and `report` is the reviewer's reply.

```python
from coursekit.protocol_examples import review_over_a2a
case = {"team_size": 3, "independent_deploys": False}
proposal = "modular monolith"
agent, report = await review_over_a2a(case, proposal)
print(f"{agent} reviewed '{proposal}': {report}")
assert report.startswith("PASS"), "The reviewer did not accept the proposal."
```

## What you should see

`Architecture reviewer reviewed 'modular monolith': PASS: course policy expects modular monolith.`

## Try it

## Takeaway

**MCP** connects an application to *tools*. **A2A** connects it to other *agents*, which can have their own models, tools, and rules. This reviewer checks agreement with a simplified policy, not design quality. A single request and reply does not show A2A's long-running tasks.
